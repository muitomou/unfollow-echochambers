```python
# --- Mapa de Calor (ε vs p_cut) ---
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

sns.set(style="whitegrid", context="talk")

df = pd.read_csv("../experiments/results_sweep/sweep_results.csv")
```

```python
# --- 1. Homofilia final ---
pivot_homo = df.pivot(index="epsilon", columns="p_cut", values="homophily")
plt.figure(figsize=(8,6))
sns.heatmap(pivot_homo, annot=True, cmap="viridis", fmt=".2f")
plt.title("Homofilia final (ε vs p_cut)")
plt.xlabel("Probabilidad de unfollow (p_cut)")
plt.ylabel("Tolerancia (ε)")
plt.tight_layout()
plt.show()
```

```python
# --- 2. Fragmentación (número de componentes) ---
pivot_comp = df.pivot(index="epsilon", columns="p_cut", values="n_components")
plt.figure(figsize=(8,6))
sns.heatmap(pivot_comp, annot=True, cmap="magma_r", fmt=".0f")
plt.title("Número de componentes (fragmentación estructural)")
plt.xlabel("Probabilidad de unfollow (p_cut)")
plt.ylabel("Tolerancia (ε)")
plt.tight_layout()
plt.show()
```

```python
# --- 3. Tamaño del componente más grande ---
pivot_frac = df.pivot(index="epsilon", columns="p_cut", values="frac_largest")
plt.figure(figsize=(8,6))
sns.heatmap(pivot_frac, annot=True, cmap="cividis", fmt=".2f")
plt.title("Fracción del componente más grande")
plt.xlabel("Probabilidad de unfollow (p_cut)")
plt.ylabel("Tolerancia (ε)")
plt.tight_layout()
plt.show()
```
